```python
first_names = ['Margaret', 'Dorothy', 'Helen', 'Betty', 'Ruth', 'Frances', 'Virginia', 'Mary', 'Jean', 'Alice',
               'Robert', 'James', 'John', 'William', 'Charles', 'George', 'Donald', 'Richard', 'Thomas', 'Walter',
               'Barbara', 'Patricia', 'Carol', 'Sandra', 'Linda', 'Nancy', 'Sharon', 'Donna', 'Susan', 'Karen',
               'Harold', 'Raymond', 'Arthur', 'Frank', 'Ralph', 'Lawrence', 'Roy', 'Howard', 'Eugene', 'Wayne']

last_names = ['Smith', 'Johnson', 'Williams', 'Brown', 'Jones', 'Miller', 'Davis', 'Wilson', 'Anderson', 'Taylor',
              'Thomas', 'Jackson', 'White', 'Harris', 'Martin', 'Thompson', 'Garcia', 'Martinez', 'Robinson', 'Clark',
              'Rodriguez', 'Lewis', 'Lee', 'Walker', 'Hall', 'Allen', 'Young', 'Hernandez', 'King', 'Wright']

cities = [('Ann Arbor', 'MI'), ('Ypsilanti', 'MI'), ('Saline', 'MI'), ('Plymouth', 'MI'), ('Canton', 'MI')]

random.seed(42)
users = []
for i in range(50):
    name = f"{random.choice(first_names)} {random.choice(last_names)}"
    age = random.randint(62, 85)
    city, state = random.choice(cities)
    physical_ability = random.choices(['low', 'medium', 'high'], weights=[40, 40, 20])[0]
    tech_comfort = random.choices(['low', 'medium', 'high'], weights=[50, 35, 15])[0]
    users.append((name, age, city, state, physical_ability, tech_comfort))

cursor.executemany("""
    INSERT INTO users (name, age, city, state, physical_ability, tech_comfort)
    VALUES (%s, %s, %s, %s, %s, %s)
""", users)
conn.commit()
print(f"Inserted {len(users)} users!")
```

Notes: 
Each time through the loop, it:
- Picks a random first name and last name
- Picks a random age between 62 and 85
- Picks a random city
- Picks a physical ability level (weighted so low and medium are more common than high — realistic for seniors)
- Picks a tech comfort level (weighted so low is most common — also realistic)
- Adds that user to the list

Then after all 50 are built, it inserts them all into MySQL at once.

```python
categories = ['outdoor', 'indoor', 'creative', 'social', 'fitness', 'educational']

survey_responses = []
for user_id in range(1, 51):
    preferred = random.sample(categories, random.randint(1, 3))
    preferred_categories = ','.join(preferred)
    available_weekdays = random.choice([True, False])
    available_weekends = random.choice([True, True, False])
    hours_per_week = random.randint(2, 15)
    prefers_outdoor = random.choice([True, False])
    prefers_social = random.choice([True, True, False])
    survey_responses.append((user_id, preferred_categories, available_weekdays,
                             available_weekends, hours_per_week, prefers_outdoor, prefers_social))

cursor.executemany("""
    INSERT INTO survey_responses (user_id, preferred_categories, available_weekdays,
                                  available_weekends, hours_per_week, prefers_outdoor, prefers_social)
    VALUES (%s, %s, %s, %s, %s, %s, %s)
""", survey_responses)
conn.commit()
print(f"Inserted {len(survey_responses)} survey responses!")
```

Notes:

Each response captured:

- preferred_categories — which types of hobbies they like (e.g. "outdoor, creative")
- available_weekdays / weekends — when they're free
- hours_per_week — how much time they have
- prefers_outdoor — do they want outdoor activities
- prefers_social — do they want group activities

for user_id in range(1, 51) loops thru user IDs 1 to 50, which are exactly the 50 users we inserted in the previous cell. So every user now has a matching survey response linked to them by user_id. 

- Users table = who they are
- Survey responses table = what they told us about themselves

```python
event_types = ['app_open', 'hobby_view', 'community_view', 'community_join', 'questionnaire_complete']

user_events = []
start_date = datetime(2025, 1, 1)

for user_id in range(1, 51):
    num_events = random.randint(10, 40)
    for _ in range(num_events):
        event_type = random.choices(
            event_types,
            weights=[30, 25, 20, 10, 15]
        )[0]
        hobby_id = random.randint(1, 20) if event_type in ['hobby_view', 'community_join'] else None
        community_id = random.randint(1, 20) if event_type in ['community_view', 'community_join'] else None
        days_offset = random.randint(0, 180)
        event_timestamp = start_date + timedelta(days=days_offset, hours=random.randint(8, 20))
        user_events.append((user_id, event_type, hobby_id, community_id, event_timestamp))

cursor.executemany("""
    INSERT INTO user_events (user_id, event_type, hobby_id, community_id, event_timestamp)
    VALUES (%s, %s, %s, %s, %s)
""", user_events)
conn.commit()
print(f"Inserted {len(user_events)} user events!")

cursor.close()
conn.close()
print("All done! Database is fully populated.")
```

Notes:
It's simulating 6 months of user behavior inside the app (January to June 2025). For each user, it randomly generates between 10 and 40 "events" — things a real user would do:

- app_open — they opened the app
- hobby_view — they clicked on a hobby to learn more
- community_view — they looked at a local club
- community_join — they actually joined a club
- questionnaire_complete — they finished the survey

The weights=[30, 25, 20, 10, 15] part means app_open happens most often (30%) and community_join happens least often (10%) — which is realistic. Most people browse more than they commit.

Each event also gets a random timestamp spread across those 6 months, so it looks like real usage data over time.
